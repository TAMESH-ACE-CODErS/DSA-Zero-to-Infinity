```python
a="hey there welcome to this video"
```

```python
count_dict={}
for item in a:
    if item in count_dict:
        count_dict[item]+=1
    else:
        count_dict[item]=1
```

```python
#counting with counter()
counts=Counter(a)
print(counts)
```

```python
#we can pass any hashable and iterable in count
count_list=Counter([1,2,3,4,5,6,7,8,9])
print(count_list)
```

```python
count_tuple=Counter((1,2,3,4,5,6,7))
print(count_tuple)
```

```python
animals=["dogs","gods","dogs","cats","cats","birds","birds","dogs"]
animals_count=Counter(animals)
```

```python
print(animals_count.get('dogs'))
```

```python
print(animals_count['bunnies'])
```

```python
print(animals_count.most_common())
```

```python
print(animals_count.most_common()[0])
```
