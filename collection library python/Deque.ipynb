```python
from collections import deque
items=deque([1,2,3,4,5])

print(type(items))

items.appendleft(0)
print(items)

print(items.append(6))
print(items)

print(items.extend([7,8,9]))
print(items)

print(items.extendleft([11,22,33]))
print(items)

print(items.popleft())
print(items)
```

```python
#maxlen
website=['google','askjeeves','weather.com']
last_vistited =deque(website,maxlen=3)
print(last_vistited)

print(last_vistited.appendleft('datagy.io'))
print(last_vistited)
```

```python
#accessing items in a deque
values=deque([1,2,3,4,5,6,7])
print(values[0])
print(values[-1])
# print(values[2:3]) #this gives error slicing not posible
```

```python
#how deques are different from list?
#comparing speed for appending 
# %%timeit #only for jupyter nb

a_list=[]
for i in range(1000):
    a_list.insert(0,0)
    
#%%timeit #only for jupyter nb
a_deque=deque()
for i in range(1000):
    a_deque.appendleft(0)
    
#list is kind of faster than deque
```

```python
#rotation 
more_value=deque([1,2,3,4,5])
more_value.rotate()
print(more_value)
more_value.rotate(-1)
print(more_value)
more_value.rotate(-3)
print(more_value)

```

```python
#adding deque
deque1=deque([1,2,3,4,5])
deque2=deque([6,7,8,9,10,11])
print(deque1+deque2)
```

```python
#Multiplying deques

print(deque1*2)
```

```python
#reverse
deque1.reverse()
print(deque1)
```
