```python
# how do dictionaries handle missing values 
inventory={ 'computer':1, 'TV':2, 'Phone':1}
inventory['Phone']+=1
print(inventory)
```

```python
if 'VCR' in inventory:
    inventory['VCR']+=1
else: 
    inventory['VCR']=1
```

```python
#USING DEFAULTDICT 
users={
    'Nik':'Toronto',
    'Kate':"Tabronto",
    'evan':'london',
    'kyra':'new york',
}
locations_old = {}
for person, location in users.items():
    if location in locations_old:
        locations_old(location).append(person)
    else :
        locations_old[location]=[person]
print(locations_old)
```

```python
#now with a default dicts
locations=defaultdict(list)
for person, location in users.items():
    locations[location].append(person)
print(locations)
```

```python
sales ={
    ('April','TVs',4234.52),
    ('April','Phones',423.32),
    ('April','Computers',12345.23),
    ('May','VCR',2432.23),
    ('May','TVs',1213.42),
    ('May','Phones',1234.52),
    ('June','TVs',3245.32)
}
aggregate=defaultdict(float)
for _,item ,amount in sales:
    aggregate[item]+=amount
print(aggregate)
```
