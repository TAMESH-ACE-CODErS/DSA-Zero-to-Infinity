```python
class Stack(list):
    
    def is_empty(self):
        return len(self) == 0
    
    def push(self, data):
        self.append(data)
        
    def pop(self):
        if not self.is_empty():
            # FIXED: Added the 'return' keyword so it gives you the popped item
            return super().pop()
        else:
            raise IndexError("stack empty re baba")
            
    def peek(self):
        if not self.is_empty():
            return self[-1]
        else:
            raise IndexError("Stack is empty")
            
    def size(self):
        return len(self)
    
    def insert(self, index, data):
        raise AttributeError("don't cheat")


# --- Testing the Code ---
s1 = Stack()

s1.push(20)

# FIXED: You must wrap it in print() to see the returned value
print("Peek after pushing 20:", s1.peek()) 

s1.push(10)
s1.push(30)

# FIXED: Now pop() will actually show you the item it removed (30)
print("Popped item:", s1.pop()) 

print("Peek after popping:", s1.peek())
```
