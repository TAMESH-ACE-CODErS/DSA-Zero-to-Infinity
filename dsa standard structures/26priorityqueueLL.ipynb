```python
class Node:
    def __init__(self,item=None,priority=None ,next=None):
        self.item=item
        self.priority=priority
        self.next=next
        
class PriorityQueue:
    def __init__(self):
        self.start=None
        self.item_count=0
        
    def push(self,data,priority):
        n=Node(data,priority)
        if not self.start or priority <self.start.priority:
            n.next=self.start
            self.start=n
        else:
            temp=self.start
            while temp.next and temp.next.priority <= priority:
                temp=temp.next
            n.next=temp.next
            temp.next=n

    def is_empty(self):
        return self.start==None       
            
    def pop(self):
        if self.is_empty():
            raise IndexError("priority queue is empty")
        data=self.start.item
        self.start=self.start.next
        self.item_count-=1
        return data
    
    def size(self):
        return self.item_count
    
p= PriorityQueue()
p.push("kanji",1)
p.push("Amit",4)
p.push("Arjun",3)
p.push(  "Kartik"   , -4    )
p.push(    "Bish" ,  3   )
p.push(   "List"  ,   3  )
p.push(     "obl",    2 )
p.push(    "kis" ,     1)
p.push( "kind"    ,   1  )
p.push( "kand"    ,    2 )

while not p.is_empty():
    print(p.pop())

        
```
