```python
from collections import defaultdict, deque

class Graph:
    def __init__(self, is_directed=False):
        # defaultdict(list) automatically creates an empty list for new vertices
        self.graph = defaultdict(list)
        self.is_directed = is_directed

    def add_edge(self, u, v):
        self.graph[u].append(v)
        # If it's an undirected graph, add the reverse connection as well
        if not self.is_directed:
            self.graph[v].append(u)

    def bfs(self, start_vertex):
        """
        Breadth-First Search looks at neighbors layer-by-layer.
        It uses a Queue (FIFO) to keep track of what to visit next.
        """
        visited = set()
        queue = deque([start_vertex]) # deque is O(1) for popping from the left
        visited.add(start_vertex)
        
        result = []

        while queue:
            current = queue.popleft()
            result.append(current)

            # Check all neighbors of the current vertex
            for neighbor in self.graph[current]:
                if neighbor not in visited:
                    visited.add(neighbor)
                    queue.append(neighbor)
        
        return result

    def dfs(self, start_vertex):
        """
        Depth-First Search plunges as deep as possible down one path before backtracking.
        It uses the Call Stack (Recursion) to keep track of what to visit next.
        """
        visited = set()
        result = []
        
        # We use a nested helper function so we don't have to pass 'visited' and 'result' everywhere
        def dfs_recursive(vertex):
            visited.add(vertex)
            result.append(vertex)
            
            for neighbor in self.graph[vertex]:
                if neighbor not in visited:
                    dfs_recursive(neighbor)
        
        dfs_recursive(start_vertex)
        return result

# --- Driver Code ---
if __name__ == "__main__":
    # Create an undirected graph
    g = Graph(is_directed=False)
    
    # Building a sample graph:
    #      0
    #    /   \
    #   1     2
    #  / \   / \
    # 3   4 5   6
    
    edges = [
        (0, 1), (0, 2),
        (1, 3), (1, 4),
        (2, 5), (2, 6)
    ]
    
    for u, v in edges:
        g.add_edge(u, v)
        
    print("Graph Adjacency List Mapping:")
    for node in sorted(g.graph.keys()):
        print(f"Vertex {node} -> {g.graph[node]}")
        
    print("\n--- Traversals ---")
    
    # BFS should print layer by layer: 0, then 1, 2, then 3, 4, 5, 6
    bfs_path = g.bfs(0)
    print(f"BFS Path starting from 0: {bfs_path}")
    
    # DFS should dive deep left first: 0, 1, 3, 4, 2, 5, 6
    dfs_path = g.dfs(0)
    print(f"DFS Path starting from 0: {dfs_path}")
```
