```python
class Node:
    def __init__(self, item=None,next=None):
        self.item=item
        self.next=next 
        
class Stack:
    def __init__(self,):
        self.start=None
        self.item_count=0
        
    def isEmpty(self):
        return self.start==None
    
    def push(self,data):
        n=Node(data,self.start)
        self.start=n
        # n.next=self.start.next    
        self.item_count+=1
    def pop(self):
        if self.isEmpty():
            return None
        else:
            data=self.start.item
            self.start=self.start.next
            self.item_count-=1
            return data
        
    def peek(self):
        if not self.isEmpty():
            return self.start.item
        else :
            return None
        
    def size(self):
        return self.item_count
    
s1=Stack()
s1.push(10);s1.push(20);s1.push(30)
print(s1.size())
print(s1.peek())
print(s1.pop())
print(s1.peek())
        
    
```
