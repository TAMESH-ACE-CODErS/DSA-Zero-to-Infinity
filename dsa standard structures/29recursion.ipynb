```python
#recursion function for sum of first N natural number
def f1(n):
    if n==0:
        return 0
    else:
        return n+f1(n-1)
    
print(f1(5))
```

```python
#recursive funcn sum of first N odd natural number
def f2(n):
    if n== 0:
        return 0
    return 2*n-1 + f2(n-1)

print(f2(5))
        
```

```python
#recursive funcn to find sum of first N even number
def f3(n):
    if n==0:
        return 0
    return 2*n + f3(n-1)

print(f3(5))
```

```python
#recursive funcn calculate factorial of a number
def f4(n):
    if n==0 or n==1:
        return 1 
    return n*f4(n-1)
print(f4(3))
```

```python
#recursive function  sum of square of first N natural number
def f5(n):
    if n==1 or n==0:
        return 1 
    return n**2 + f5(n-1)

print(f5(3))
```
