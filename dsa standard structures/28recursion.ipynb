```python
#write a recursive function to print first n natural number
def f1(n):
    if n>=0:
        f1(n-1)
        print(n)
    return n+1
    
print(f1(6))
```

```python
#write a recursive function to print natural number in reverse order
def f2(n):
    if n==0:
        return 0
    print(n)
    f2(n-1)
    
    
print(f2(6))
```

```python
# write a code for odd n natural number
def f3(n):
    if n>0:
        f3(n-1)
        print(2*n-1)
print(f3(5))
        
    
```

```python
#reverse order of odd number
def f4(n):
    if n>0:
        print(2*n-1)
        f3(n-1)
        
print(f3(5))
```

```python
#even number 
def f5(n):
    if n>0:
        f5(n-1)
        print(2*n)
print(f5(10))
```
