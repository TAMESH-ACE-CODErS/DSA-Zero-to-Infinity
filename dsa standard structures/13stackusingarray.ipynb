```python
class Stack:
    def __init__(self):
        self.item=[]
        
    def is_empty(self):
        return len(self.item)==0
    
    def push(self,data):
        self.item.append(data)
        
    def pop(self):
        if not self.is_empty():
            self.item.pop()
        else:
            return ; #do nothing 
    
    def peek(self):
        if not self.is_empty():
            return self.item[-1]
        else:
            return ;
    
    def size(self):
        return len(self.item)
    
s1=Stack()
s1.push(10);
s1.push(20)
s1.push(30)
s1.pop()
s1.peek()
```
