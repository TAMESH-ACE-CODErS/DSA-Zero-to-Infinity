```python
class Queue:
    
    def __init__(self):
        self.items=[]
        # self.front=None
        # self.rear=None
        #currently maintaining front and rear is a burden currently
        
    def is_Empty(self):
        return len(self.items)==0
    
    def enqueue(self,data):
        self.items.append(data)
        # self.front=self.front+1
        
    def dequeue(self):
        self.items.pop(0)
        if not self.is_Empty():
            self.items.pop(0)
        else:
            raise IndexError("Queue Underflow")
        
    def get_front(self):
        if not self.is_Empty():
            return self.items[-0]
        else:
            raise IndexError("Queue Underflow")
        
    def get_rear(self):
        if not self.is_Empty():
            return self.items[-1]
        else:
            raise IndexError ("Queue Underflow")
    
    def size(self):
        return len(self.items)
    
q1=Queue()
q1.enqueue(10)
q1.enqueue(20)
try:
    print(q1.get_front())
    print(q1.get_rear())
except IndexError as e:
    print(e.args[0])
print()
    

        
    
```
