```python
class Node:
    def __init__(self, item=None, next=None):
        self.item = item
        self.next = next
        
class CLL:
    def __init__(self, last=None):
        self.last = last 
        
    def isEmpty(self):
        return self.last == None
    
    def insert_at_start(self, data):
        n = Node(data)
        if self.isEmpty():
            n.next = n
            self.last = n
        else:
            n.next = self.last.next
            self.last.next = n

    def insert_at_last(self, data):
        n = Node(data)
        if self.isEmpty():
            n.next = n
            self.last = n
        else:
            n.next = self.last.next
            self.last.next = n
            self.last = n # The only difference from insert_at_start

    def search(self, data):
        if self.isEmpty(): return None
        temp = self.last.next
        while temp != self.last:
            if temp.item == data: return temp
            temp = temp.next
        return temp if temp.item == data else None
    
    def insert_after(self, temp, data):
        if temp is not None:
            n = Node(data, temp.next)
            temp.next = n
            if temp == self.last:
                self.last = n
    
    def delete_first(self):
        if not self.isEmpty():
            if self.last.next == self.last:
                self.last = None
            else:
                self.last.next = self.last.next.next

    def delete_last(self):
        if not self.isEmpty():
            if self.last.next == self.last:
                self.last = None
            else:
                temp = self.last.next
                while temp.next != self.last:
                    temp = temp.next
                temp.next = self.last.next
                self.last = temp
                
    def delete_item(self, data):
        if self.isEmpty():
            return

        # Scenario 1: Only one node in the list
        if self.last.next == self.last:
            if self.last.item == data:
                self.last = None
            return

        # Scenario 2: Deleting the FIRST node (the Head)
        # Remember: self.last.next is the head
        if self.last.next.item == data:
            self.delete_first()
            return

        # Scenario 3: Deleting a node in the middle or at the end
        temp = self.last.next
        while temp.next != self.last.next:
            if temp.next.item == data:
                # If the node to delete is actually the LAST node
                if temp.next == self.last:
                    self.last = temp
                
                # Bridge the gap
                temp.next = temp.next.next
                break
            temp = temp.next

    def __iter__(self):
        if self.last == None:
            return CLLIterator(None)
        return CLLIterator(self.last.next)
                    
class CLLIterator:
    def __init__(self, start):
        self.current = start
        self.start = start
        self.first_pass = True # To handle the circular nature
        
    def __iter__(self):
        return self

    def __next__(self):
        if self.current == None:
            raise StopIteration
        
        # If we are back at start and it's NOT the first time, stop
        if self.current == self.start and not self.first_pass:
            raise StopIteration
        
        self.first_pass = False
        data = self.current.item
        self.current = self.current.next
        return data

# --- TESTING THE IMPLEMENTATION ---
cll = CLL()
cll.insert_at_start(10) # List: 10
cll.insert_at_start(20) # List: 20, 10
cll.insert_at_last(30)  # List: 20, 10, 30
cll.insert_after(cll.search(10), 15) # List: 20, 10, 15, 30

cll.delete_first() # Removes 20
cll.delete_last()  # Removes 30

for x in cll:
    print(x, end=" ") # Expected: 10 15
```
